```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

## Relational Plot

```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
#scatter plot:

sns.scatterplot(data=tips, x='total_bill', y='tip', hue='sex', style='time', size='size')
```

```python
#relplot (figure level function)

sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter', hue='sex', style='time', size='size')
```

```python
#line plot:

gap = px.data.gapminder()
gap
```

```python
temp = gap[gap['country']=='India']
temp
```

```python
#axis level function

sns.lineplot(data=temp, x='year', y='lifeExp')
```

```python
sns.relplot(data=temp, x='year', y='lifeExp', kind='line')
```

```python
#new country temp:

temp = gap[gap['country'].isin(['India', 'Pakistan', 'China'])]
temp
```

```python
sns.relplot(kind='line', data=temp, x='year', y='lifeExp', hue='country')
```

```python
#using axes function:

sns.lineplot(data=temp, x='year', y='lifeExp', hue='country')
```

```python
temp = gap[gap['country'].isin(['India', 'Brazil', 'Germany'])]
temp
```

```python
sns.relplot(kind='line', data=temp, x='year', y='lifeExp', style='continent', hue='country')
```
