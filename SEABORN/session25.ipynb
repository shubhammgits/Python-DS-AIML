```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

## Relational Plot

```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
#scatter plot:

sns.scatterplot(data=tips, x='total_bill', y='tip', hue='sex', style='time', size='size')
```

```python
#relplot (figure level function)

sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter', hue='sex', style='time', size='size')
```

```python
#line plot:

gap = px.data.gapminder()
gap
```

```python
temp = gap[gap['country']=='India']
temp
```

```python
#axis level function

sns.lineplot(data=temp, x='year', y='lifeExp')
```

```python
sns.relplot(data=temp, x='year', y='lifeExp', kind='line')
```

```python
#new country temp:

temp = gap[gap['country'].isin(['India', 'Pakistan', 'China'])]
temp
```

```python
sns.relplot(kind='line', data=temp, x='year', y='lifeExp', hue='country')
```

```python
#using axes function:

sns.lineplot(data=temp, x='year', y='lifeExp', hue='country')
```

```python
temp = gap[gap['country'].isin(['India', 'Brazil', 'Germany'])]
temp
```

```python
sns.relplot(kind='line', data=temp, x='year', y='lifeExp', style='continent', hue='country')
```

```python
#facet plot:

tips.head()
```

```python
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter', col='smoker', row='sex')
```

```python
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter', col='sex', row='day')
```

```python
#col wrap:
gap
```

```python
sns.relplot(kind='scatter', data=gap, x='lifeExp', y='gdpPercap', col='year', col_wrap=3)
```

## Distribution Plot

```python
#plotting univarient histogram:

sns.histplot(data=tips, x='total_bill')
```

```python
#figure level:

sns.displot(data=tips, x='total_bill', kind='hist')
```

```python
#bins parameter:
sns.displot(data=tips, x='total_bill', kind='hist', bins=2)
```

```python
#categorical columns:

sns.displot(data=tips, kind='hist', x='day')
```

```python
#hue parameter:

sns.displot(data=tips, x='tip', kind='hist', hue='sex', element='step')
```

```python
titanic=sns.load_dataset('titanic')
titanic
```

```python
sns.displot(data=titanic, x='alive', kind='hist', element='step', hue='sex')
```

```python
#faceting:

sns.displot(data=titanic, x='alive', kind='hist', col='sex')
```

```python
#kde plot:

sns.kdeplot(data=tips, x='total_bill')
```

```python
sns.displot(data=tips, x='total_bill', kind='kde', hue='sex', fill=True)
```
