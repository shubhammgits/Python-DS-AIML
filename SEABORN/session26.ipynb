## Advanced Seaborn

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
#import dataset:

tips=sns.load_dataset('tips')
iris=sns.load_dataset('iris')
```

```python
tips.head()
```

## Categorical Plots

```python
#categorical scatter plot:

#stripplot:

sns.stripplot(data=tips, x='day', y='total_bill', hue='day', jitter=False)
```

```python
sns.catplot(data=tips, x='total_bill', y='day', kind='strip',  hue='day')
```

```python
sns.catplot(data=tips, y='total_bill', x='day', kind='strip',  hue='day', jitter=0.3)
```

```python
#swarmplot:

sns.catplot(data=tips, x='day', y='total_bill', kind='swarm', hue='day')
```

## Categorical Distribution Plot

```python
#boxplot:

sns.boxplot(data=tips, x='sex', y='total_bill', hue='sex')
```

```python
sns.catplot(data=tips, kind='box', x='day', y='total_bill', hue='day')
```

```python
#violin plot:;

sns.violinplot(data=tips, x='day', y='total_bill', hue='day')
```

## Categorical Estimate Plot

```python
#bar plot:

sns.barplot(data=tips, x='sex', y='total_bill', hue='smoker')
```

```python
sns.catplot(data=tips, kind='bar', x='sex', y='total_bill', hue='smoker', estimator=np.median)
```

```python
#point plot:

sns.barplot(data=tips, x='sex', y='total_bill')
```

```python
#point plot:

sns.pointplot(data=tips, x='sex', y='total_bill', hue='smoker')
```

```python
#countplot:

sns.countplot(data=tips, x='sex')
```

```python
sns.countplot(data=tips, x='day', hue='sex')
```

```python
#faceting using catplot:

sns.catplot(data=tips, x='sex', y='total_bill', col='smoker', kind='box', row='time')
```

```python
#reg plot:

sns.regplot(data=tips, x='total_bill', y='tip')
```

```python
sns.lmplot(data=tips, x='total_bill', y='tip', hue='sex')
```

```python
#resdplot:

sns.residplot(data=tips, x='total_bill', y='tip')
```

```python
#facetplot:

sns.catplot(data=tips, x='sex', y='total_bill', kind='violin', hue='sex', col='day', row='time')
```

```python
#using facetgrid:

g = sns.FacetGrid(data=tips,col='day', row='time')

g.map(sns.violinplot, 'sex', 'total_bill')
```
