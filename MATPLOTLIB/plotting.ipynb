# PLOTTING USING MATPLOTLIB

```python
#importing the library:

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#plotting a simple function:

price=[48000,54000,57000,49000,47000,45000,49000]
year=[2015,2016,2017,2018,2019,2020,2021]
plt.plot(year,price)
```

```python
#label title:

plt.plot(batsman['index'], batsman['V Kohli'])
plt.plot(batsman['index'], batsman['RG Sharma'])

plt.title('Rohit Sharma vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
#colors(hex) and line(width and style) and marker(size)

plt.plot(batsman['index'], batsman['V Kohli'], color = 'red', linestyle = 'dotted', linewidth = 2, marker='+', markersize=14)
plt.plot(batsman['index'], batsman['RG Sharma'], color = '#FC00D6', linestyle='dashed', linewidth = 3, marker='D', markersize=15)

plt.title('Rohit Sharma vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
#legend --> location

plt.plot(batsman['index'], batsman['V Kohli'], color='red', linewidth=3, marker='+', markersize=10, linestyle='dotted', label='virat kohli')
plt.plot(batsman['index'], batsman['RG Sharma'], color='blue', linewidth=3, marker='D', linestyle = 'dotted', label='rohit sharma')

plt.title('V Kohli v/s Rohit Sharma Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')

# plt.legend(loc='right')
plt.legend()
```

```python
#limiting axes:

price = [48000,50000,49000,47000,45000,4500000]
year= [2015,2016,2017,2018,2019,2020]

plt.plot(year,price)
```

```python
price = [48000,50000,49000,47000,45000,450000]
year= [2015,2016,2017,2018,2019,2020]

plt.plot(year,price)

plt.ylim(40000,50000)
plt.xlim(2017,2019)
```

```python
#grid:

plt.plot(batsman['index'], batsman['V Kohli'], color='red', linewidth=3, marker='+', markersize=10, linestyle='dotted', label='virat kohli')
plt.plot(batsman['index'], batsman['RG Sharma'], color='blue', linewidth=3, marker='D', linestyle = 'dotted', label='rohit sharma')

plt.title('V Kohli v/s Rohit Sharma Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')

# plt.legend(loc='right')
plt.legend()

plt.grid()

plt.show()
```

## BAR GRAPH

```python
colors=['red', 'blue', 'green', 'yellow', 'pink']
children = [10,20,40,10,20]

plt.bar(colors,children)
```

```python
plt.barh(colors,children, color='black')
```

```python
#multiple bar charts

df = pd.read_csv('datasets/batsman_season_record.csv')
df
```

```python
plt.bar(df['batsman'], df['2015'])
plt.bar(df['batsman'], df['2016'])
```

```python
plt.bar(np.arange(df.shape[0]) - 0.2, df['2015'], width=0.2)
plt.bar(np.arange(df.shape[0]), df['2016'], width=0.2)
plt.bar(np.arange(df.shape[0]) + 0.2, df['2017'], width=0.2)

plt.xticks(np.arange(df.shape[0]), df['batsman'])
plt.show()
```

```python
#problem:

children = [10,20,40,10,30]
colors = ['re red red red', 'blue blue blue blue blue', 'green green green green green', 'pink pink pink pink pink', 'yellow yellow yellow yellow yellow yellow']

plt.bar(colors, children)
plt.xticks(rotation = 'vertical')

plt.show()
```
