### 3D Line Plot

### 3D Surface Plots

```python
#f(x,y) = x^2 + y^2 

x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)

xx,yy=np.meshgrid(x,y)

z = xx**2 + yy**2
z.shape
```

```python
fig = plt.figure(figsize=(12,8))
ax= plt.subplot(projection= '3d')

p = ax.plot_surface(xx,yy,z, cmap = 'viridis')
fig.colorbar(p)
```

```python
#another example:

z = np.sin(xx) + np.cos(yy)

fig = plt.figure(figsize = (12,8))
ax = plt.subplot(projection='3d')
p=ax.plot_surface(xx,yy,z, cmap='Grays')
fig.colorbar(p)
```

## Contour Plots

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)

xx,yy=np.meshgrid(x,y)

z = xx**2 + yy**2
z.shape

fig = plt.figure(figsize=(12,8))
ax= plt.subplot(projection= '3d')

p = ax.plot_surface(xx,yy,z, cmap = 'viridis')
fig.colorbar(p)
```

```python
#contour plotting of above graph:

fig = plt.figure(figsize=(12,8))
ax= plt.subplot()

p = ax.contour(xx,yy,z, cmap = 'viridis')
fig.colorbar(p)
```

```python
#contourf:

fig = plt.figure(figsize=(12,8))
ax= plt.subplot()

p = ax.contourf(xx,yy,z, cmap = 'viridis')
fig.colorbar(p)
```
