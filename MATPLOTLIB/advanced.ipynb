# Advanced Matplotlib

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### Colored Scatterplots

```python
iris = pd.read_csv('datasets/iris.csv')
iris.head()
```

```python
plt.scatter(iris['SepalLengthCm'], iris['PetalLengthCm'])
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
```

```python
iris['Species'] = iris['Species'].replace({'Iris-setosa':0, 'Iris-versicolor':1, 'Iris-virginica':2})
iris.sample(5)
```

```python
plt.scatter(iris['SepalLengthCm'], iris['PetalLengthCm'], c=iris['Species'], cmap='winter')
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')

plt.colorbar()
```

## Plot size

```python
plt.figure(figsize=(15,7))

plt.scatter(iris['SepalLengthCm'], iris['PetalLengthCm'], c=iris['Species'], cmap='winter')
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')

plt.colorbar()
```

## Annotations

```python
batters = pd.read_csv('datasets/batter.csv')

sample = batters.head(100).sample(25, random_state=5)

sample
```

```python
plt.scatter(sample['avg'], sample['strike_rate'])
```

```python
x = [1,2,3,4]
y= [5,6,7,8]

plt.scatter(x,y)

plt.text(1,5,'Point 1')
plt.text(2,6,'Point 2')
plt.text(3,7,'Point 3', fontdict={'size':12, 'color':'brown'})
```

```python
#for dataset:
plt.figure(figsize = (10,10))
plt.scatter(sample['avg'], sample['strike_rate'], s=sample['runs'])

for i in range(sample.shape[0]):
    plt.text(sample['avg'].values[i], sample['strike_rate'].values[i], sample['batter'].values[i])
```

## Horizontal/Vertical Lines

```python
plt.figure(figsize = (10,10))
plt.scatter(sample['avg'], sample['strike_rate'], s=sample['runs'])

plt.axhline(130, color='red')
plt.axvline(30, color='red')

for i in range(sample.shape[0]):
    plt.text(sample['avg'].values[i], sample['strike_rate'].values[i], sample['batter'].values[i], fontdict={'size':15})
```

## 3D Plots

### 3D Surface Plots

```python
#f(x,y) = x^2 + y^2 

x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)

xx,yy=np.meshgrid(x,y)

z = xx**2 + yy**2
z.shape
```

```python
fig = plt.figure(figsize=(12,8))
ax= plt.subplot(projection= '3d')

p = ax.plot_surface(xx,yy,z, cmap = 'viridis')
fig.colorbar(p)
```

```python
#another example:

z = np.sin(xx) + np.cos(yy)

fig = plt.figure(figsize = (12,8))
ax = plt.subplot(projection='3d')
p=ax.plot_surface(xx,yy,z, cmap='Grays')
fig.colorbar(p)
```

## Contour Plots

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)

xx,yy=np.meshgrid(x,y)

z = xx**2 + yy**2
z.shape

fig = plt.figure(figsize=(12,8))
ax= plt.subplot(projection= '3d')

p = ax.plot_surface(xx,yy,z, cmap = 'viridis')
fig.colorbar(p)
```
