```python
#handling dates:
pd.read_csv('datasets/IPL Matches 2008-2020.csv', parse_dates = ['date']).info()
```

```python
pd.read_csv('datasets/parse_date.csv', parse_dates=[['day','month', 'date', 'year']])
```

```python
#converters:

def rename(name):
    if name == 'Royal Challengers Bangalore':
        return 'RCB'
    else:
        return name
    
pd.read_csv('datasets/IPL Matches 2008-2020.csv', converters = {'team1' : rename})
```

```python
#na_values parameter:
pd.read_csv('datasets/aug_train.csv', na_values=['Male'])
```

```python
#loading a huge dataset in chunks:

dfs = pd.read_csv('datasets/aug_train.csv', chunksize=5000)
```

```python
for chunks in dfs:
    print(chunks.shape)
```
