## Pivot Table

```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
df = sns.load_dataset('tips')
df.head()
```

```python
#avf total bill gender wise:

df.groupby('sex')[['total_bill']].mean()
```

```python
#avg bill paid by somker males and smoker females:
df.groupby(['sex', 'smoker'])[['total_bill']].mean().unstack()
```

```python
#using pivot:

df.pivot_table(index = 'sex', columns = 'smoker', values = 'total_bill')
```

```python
df.pivot_table(index='sex', columns='smoker', values='total_bill', aggfunc='count')
```

```python
df.pivot_table(index='sex', columns='smoker', aggfunc='count')
```

```python
#multidimensional:
df.pivot_table(index=['sex', 'smaoker'], columns=['day', 'time'], values='total_bill')
```
